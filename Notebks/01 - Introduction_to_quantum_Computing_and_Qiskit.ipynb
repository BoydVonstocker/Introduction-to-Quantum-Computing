```python
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector
from qiskit.visualization import plot_bloch_multivector, plot_histogram, array_to_latex
from qiskit_aer import AerSimulator
```

## Why quantum computing starts with a qubit

A classical bit is either `0` or `1`. A **qubit** is a two-level quantum system whose state can be any unit vector in a two-dimensional complex vector space.

The standard computational basis is

$$
|0\rangle = \begin{pmatrix} 1 \\ 0 \end{pmatrix}, \qquad
|1\rangle = \begin{pmatrix} 0 \\ 1 \end{pmatrix}.
$$

A general pure qubit state is

$$
|\psi\rangle = \alpha|0\rangle + \beta|1\rangle = \begin{pmatrix} \alpha \\ \beta \end{pmatrix},
$$

with complex amplitudes $\alpha,\beta$ that satisfy the normalization condition

$$
|\alpha|^2 + |\beta|^2 = 1.
$$

$|\alpha|^2$ is the probability of measuring `0`. $|\beta|^2$ is the probability of measuring `1`.


## Dirac notation in one minute

- $|\psi\rangle$ is a column vector (a *ket*).
- $\langle\psi|$ is its conjugate transpose (a *bra*).
- The inner product $\langle\phi|\psi\rangle$ is a complex number.
- The projector $|\psi\rangle\langle\psi|$ is a matrix.

Two states are **orthogonal** when $\langle\phi|\psi\rangle = 0$. The computational basis states satisfy $\langle 0|1\rangle = 0$.


## The default Qiskit qubit

In Qiskit, a new qubit starts in $|0\rangle$. The cell below builds a one-qubit circuit with no gates and prints its statevector.


```python
qc = QuantumCircuit(1)
qc.draw("mpl")
```

```python
sv = Statevector.from_instruction(qc)
display(array_to_latex(sv, prefix=r"$|\psi\rangle = $"))
print("amplitudes:", sv.data)
print("probabilities:", sv.probabilities_dict())
plot_bloch_multivector(sv)
```

## Preparing $|1\rangle$

Apply an $X$ gate (the quantum NOT) to flip $|0\rangle$ into $|1\rangle$:

$$
X|0\rangle = |1\rangle.
$$


```python
qc = QuantumCircuit(1)
qc.x(0)
display(qc.draw("mpl"))

sv = Statevector.from_instruction(qc)
display(array_to_latex(sv, prefix=r"$|\psi\rangle = $"))
plot_bloch_multivector(sv)
```

## A first superposition

The Hadamard gate $H$ sends $|0\rangle$ to an equal superposition:

$$
H|0\rangle = \frac{|0\rangle + |1\rangle}{\sqrt{2}} \equiv |+\rangle.
$$

Both measurement outcomes are then equally likely.


```python
qc = QuantumCircuit(1)
qc.h(0)
display(qc.draw("mpl"))

sv = Statevector.from_instruction(qc)
display(array_to_latex(sv, prefix=r"$|+\rangle = $"))
print(sv.probabilities_dict())
plot_bloch_multivector(sv)
```

## Measuring that superposition

Statevectors give the ideal amplitudes. A simulator with shots samples the measurement distribution. Add a classical bit, measure, and run 1024 shots.


```python
qc = QuantumCircuit(1, 1)
qc.h(0)
qc.measure(0, 0)
display(qc.draw("mpl"))

sim = AerSimulator()
counts = sim.run(qc, shots=1024).result().get_counts()
print(counts)
plot_histogram(counts)
```

## Two qubits, briefly

Two qubits live in a 4-dimensional space with basis

$|00\rangle,\ |01\rangle,\ |10\rangle,\ |11\rangle$.

Qiskit prints bitstrings with qubit 0 on the **right** (little-endian). Keep that in mind when you read labels such as `01`.


```python
qc = QuantumCircuit(2)
qc.x(0)   # flip qubit 0 only
display(qc.draw("mpl"))
sv = Statevector.from_instruction(qc)
print(sv.probabilities_dict())
display(array_to_latex(sv, prefix=r"$|\psi\rangle = $"))
```
